```python
import gzip
import numpy as np
import matplotlib
#matplotlib.use('Agg')
%matplotlib inline  
matplotlib.rcParams.update({'font.size': 12})
import matplotlib.pyplot as plt

def read3d(fn):
    with gzip.open(fn, 'rb') as f:
        fread = f.read()
        nv = np.frombuffer(fread, dtype=np.int32,  count=1, offset=0)[0]
        vx = np.frombuffer(fread, dtype=np.double, count=nv, offset=4)
        vy = np.frombuffer(fread, dtype=np.double, count=nv, offset=4+nv*8*1)
        vz = np.frombuffer(fread, dtype=np.double, count=nv, offset=4+nv*8*2)
        arr  = np.frombuffer(fread, dtype=np.double, offset=4+nv*8*3).reshape([-1,1+nv])
        #arr  = np.frombuffer(fread, dtype=np.double, offset=4+nv*8*3)
        #print(arr.shape)
        
        #print (arr)
    return vx,vy,vz,arr
         
P3AVG="./fZ400/v064p01/p3savg.dat"
P3AVG="./fZ600a/v128p01/p3savg.dat"
plt.figure(figsize=(10, 4))

TAG=["0_9","1_0","1_1","1_2","1_3"]
for t in TAG:
  FN=f"./fZ600a{t}/v128p01/p3savg.dat" 
  vx,vy,vz,d = read3d(FN)
  plt.plot(vz,d[-1,1:], ".-", label=f"a={t}")
plt.legend()
plt.xlabel(f"V")
plt.show()

```

```python
import gzip
import numpy as np
import matplotlib
#matplotlib.use('Agg')
%matplotlib inline  
matplotlib.rcParams.update({'font.size': 12})
import matplotlib.pyplot as plt

def read1d(fn):
    with gzip.open(fn, 'rb') as f:
        fread = f.read()
        nv = np.frombuffer(fread, dtype=np.int32,  count=1, offset=0)[0]
        vz = np.frombuffer(fread, dtype=np.double, count=nv, offset=4)
        arr  = np.frombuffer(fread, dtype=np.double, offset=4+nv*8).reshape([-1,1+nv])
        #arr  = np.frombuffer(fread, dtype=np.double, offset=4+nv*8*3)
        #print(arr.shape)
        
        #print (arr)
    return vx,vy,vz,arr
         
plt.figure(figsize=(10, 4))

TAG=["0_9","1_0","1_1","1_2","1_3"]
for t in TAG:
  FN=f"../../nuosc1d/ben_avg/fZ600a{t}/v128/p3savg.dat" 
  vx,vy,vz,d = read1d(FN)
  plt.plot(vz,d[-1,1:], ".-", label=f"a={t}")
plt.hlines(0.5,-1,1,"k",":")
plt.legend()
plt.ylim([0,1])
plt.xlim([-1,1])
plt.xlabel(f"V")
plt.show()

```
